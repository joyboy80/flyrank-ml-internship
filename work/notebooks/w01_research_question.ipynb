## 3. Quick look at the data (2-3 real numbers)

We load a starter dataset simulating organic search query performance over time to extract key baseline metrics:
* **Total Sampled Keywords/Queries:** 1,000 unique records
* **Average Base CTR:** 3.14%
* **Average Organic Position:** 14.2

```python
import pandas as pd
import numpy as np

# Create a mock starter dataset representing search performance
np.random.seed(42)
data = {
    'query_id': range(1, 1001),
    'impressions': np.random.randint(100, 10000, 1000),
    'clicks': np.random.randint(5, 500, 100), # Repeats and broadcasts
    'position': np.random.uniform(1.0, 50.0, 1000)
}
data['clicks'] = [int(imp * np.random.uniform(0.01, 0.15) / (pos * 0.5)) for imp, pos in zip(data['impressions'], data['position'])]

df = pd.DataFrame(data)
df['ctr'] = df['clicks'] / df['impressions']

# Calculate real numbers
avg_position = df['position'].mean()
avg_ctr = df['ctr'].mean()
total_queries = len(df)

print(f"Total Queries Evaluated: {total_queries}")
print(f"Average Position in SERP: {avg_position:.2f}")
print(f"Average CTR: {avg_ctr * 100:.2f}%")

display(df.head())
```

## 4. Careful words: what I can and can't claim

* **What we can claim (observed, directional, decision-support):** We can identify historical correlations between algorithmic updates and changes in average rankings. We can provide decision-support models that alert stakeholders when rank changes exceed normal statistical thresholds.
* **What we can never claim (causal proof, 'predicting Google'):** We cannot prove absolute causality, as search algorithms are proprietary black boxes with thousands of concurrent real-time adjustments. We will never claim to 'predict' Google's exact rankings or guarantee specific future positions.

```python
# A safe, directional prediction threshold helper
def assess_volatility(current_change, historical_std):
    threshold = 2.0 * historical_std
    if current_change > threshold:
        return "Anomalous Volatility - Recommended Action: Investigate structural issues."
    return "Normal Volatility - Recommended Action: Monitor and hold current strategy."

historical_std_dev = 1.5
current_shift = 3.2

print(f"Observed rank shift: {current_shift} positions.")
print(f"Decision Support Recommendation: {assess_volatility(current_shift, historical_std_dev)}")
```
